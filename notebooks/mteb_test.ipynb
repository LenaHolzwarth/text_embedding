```python
class ModelCard():
    def __init__(self, name: str):
        self.model_name = name
        self.language = ""
        self.base_model_revision = 1.0
        
```

```python
# implement the tf-idf class for mteb to use
class Tfidf():
    
    def __init__(self):
        self.model_card_data = ModelCard("Tfidf")
        #self.model_name = 'tfidf'
    
    def encode(self, sentences: list[str], **kwargs: Any
    ) -> torch.Tensor | np.ndarray:
        """Encodes the given sentences using the encoder.

        Args:
            sentences: The sentences to encode.
            **kwargs: Additional arguments to pass to the encoder.

        Returns:
            The encoded sentences.
        """
        # initialize the model
        vectorizer = TfidfVectorizer()
        # fit on data
        sent_vec = vectorizer.fit_transform(sentences)
        # transform sparse matrix to numpy array
        sent_np = sent_vec.toarray()

        return sent_np
```
