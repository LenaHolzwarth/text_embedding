#### 1. compare self-made get_vocab with sklearn tfidf version

```python
import datasets
import json
from numpyencoder import NumpyEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from mteb.evaluation.evaluators.utils import get_vocab
```

```python
# load dataset
arxiv_data = datasets.load_dataset("mteb/arxiv-clustering-p2p", revision="a122ad7f3f0291bf49cc6f4d32aa80929df69d5d")["test"]
biorxiv_data = datasets.load_dataset("mteb/biorxiv-clustering-p2p", revision="f5dbc242e11dd8e24def4c4268607a49e02946dc")["test"]
```

```python
arxiv_sents = [x for split in arxiv_data for x in split["sentences"]]
biorxiv_sents = [split["sentences"] for split in biorxiv_data]
```

```python
# compute vocab with get_vocab
vocab_selfmade = get_vocab(biorxiv_sents)
```

```python
# compute vocab with sklearn 
# initialize the model
vectorizer = TfidfVectorizer()
# fit on data
vectorizer_fit = vectorizer.fit(biorxiv_sents)

vocab_sklearn = vectorizer_fit.vocabulary_
```

### 2. Compare vocabs for different settings of get_vocab

#### a) vary max_features and ngram_range
max_features ignores infrequent words in the corpus (independent of their distribution)

range: [1000, 2000, 5000, 10000, unlimited]

```python
ngram_range_ = [(1,1), (1,2), (2,2)]
max_df_ = [1.0, 0.2, 0.1, 0.05, 0.02, 0.01]
min_df_ = [1, 0.01, 0.02, 0.05, 0.1]
max_features_ = [None]
```

```python
# Arxiv data
dataset_name = "arxiv"
for ngram_range in ngram_range_:
    for max_df in max_df_:
        for min_df in min_df_:
            if min_df >= max_df and not min_df==1:
                continue
            for max_features in max_features_:
                vocab = get_vocab(biorxiv_sents, 
                                  ngram_range = ngram_range,
                                  max_df = max_df,
                                  min_df = min_df,
                                  max_features = max_features)
                
                with open(f"text_embedding/MTEB/vocabs/{dataset_name}/ngram_{ngram_range[0]}_{ngram_range[1]}_maxdf_{int(max_df)}-{int(round(max_df%1, 2)*10)}{int(round(max_df%1, 2)*100)}_mindf_{int(min_df)}-{int(round(min_df%1, 2)*10)}{int(round(min_df%1, 2)*100)}_maxfeatures_{max_features}.json", "w") as fp:
                    json.dump(vocab , fp, indent = 4, ensure_ascii=False, cls=NumpyEncoder) 
                
```

```python
# Biorxiv data
dataset_name = "biorxiv"
for ngram_range in ngram_range_:
    for max_df in max_df_:
        for min_df in min_df_:
            if min_df >= max_df and not min_df==1:
                continue
            for max_features in max_features_:
                vocab = get_vocab(biorxiv_sents, 
                                  ngram_range = ngram_range,
                                  max_df = max_df,
                                  min_df = min_df,
                                  max_features = max_features)
                
                with open(f"text_embedding/MTEB/vocabs/{dataset_name}/ngram_{ngram_range[0]}_{ngram_range[1]}_maxdf_{int(max_df)}-{int(round(max_df%1, 2)*10)}{int(round(max_df%1, 2)*100)}_mindf_{int(min_df)}-{int(round(min_df%1, 2)*10)}{int(round(min_df%1, 2)*100)}_maxfeatures_{max_features}.json", "w") as fp:
                    json.dump(vocab , fp, indent = 4, ensure_ascii=False, cls=NumpyEncoder) 
                
```
