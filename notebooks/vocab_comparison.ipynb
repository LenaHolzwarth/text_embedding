Evaluate different vocab functions and what types of words they select

#### 1. compare self-made get_vocab with sklearn tfidf version

```python
import datasets
from sklearn.feature_extraction.text import TfidfVectorizer
from mteb.evaluation.evaluators.utils import get_vocab
```

```python
# load dataset
arxiv_data = datasets.load_dataset("mteb/arxiv-clustering-p2p", revision="a122ad7f3f0291bf49cc6f4d32aa80929df69d5d")["test"]
biorxiv_data = datasets.load_dataset("mteb/biorxiv-clustering-p2p", revision="f5dbc242e11dd8e24def4c4268607a49e02946dc")["test"]
```

```python
arxiv_sents = [x for split in arxiv_data for x in split["sentences"]]
biorxiv_sents = [split["sentences"] for split in biorxiv_data]
```

```python
# compute vocab with get_vocab
vocab_selfmade = get_vocab(biorxiv_sents)
```

```python
# compute vocab with sklearn 
# initialize the model
vectorizer = TfidfVectorizer()
# fit on data
vectorizer_fit = vectorizer.fit(biorxiv_sents)

vocab_sklearn = vectorizer_fit.vocabulary_
```

```python
len(vocab_sklearn)
```

```python
vocab_sklearn_lower = [word for word in sorted(list(vocab_sklearn.keys())) if word.islower()]
```

```python
len(vocab_sklearn_lower)
```

```python
for word in vocab_sklearn_lower:
    if not word in vocab_selfmade:
        print(word)
```
