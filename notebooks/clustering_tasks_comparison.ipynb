### Compare MTEB Clustering to MTEB ClusteringFast

```python
import mteb
import src
```

```python
# set up model
glove = "sentence-transformers/average_word_embeddings_glove.6B.300d"
model = mteb.get_model(glove)

# directory for results
results_folder = "../MTEB/clustering_comparison"
```

#### Arxiv 

```python
evaluation = mteb.MTEB(tasks = mteb.get_tasks(tasks=["ArxivClusteringP2P"]))
results = evaluation.run(model, output_folder=results_folder,
                         encode_kwargs = {"ngram_range": (1,1), 
                                          "max_df": 1.0, 
                                          "min_df": 1, 
                                          "max_features": None})
```

```python
evaluation = mteb.MTEB(tasks = mteb.get_tasks(tasks=["ArXivHierarchicalClusteringS2S"]))
results = evaluation.run(model, output_folder=results_folder,
                         encode_kwargs = {"ngram_range": (1,1), 
                                          "max_df": 1.0, 
                                          "min_df": 1, 
                                          "max_features": None})
```
